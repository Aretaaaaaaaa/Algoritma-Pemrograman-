```python
# no 1
#list buah = ["apel, mangga, jeruk, anggur, pisang"]
data = ["apel", "mangga", "jeruk", "anggur", "pisang"]

# a tambahkan senangka di akhir list
data.append("semangka")
print(f"data setelah ditambah = {data}\t")

# b sisipkan 'durian' di anatara 'jeruk' dan 'anggur'
data.insert(3, "durian")
print(f"data setelah disisipkan : {data}")

# c hapus mangga dari ist
data.remove("mangga")
print(f"setelah dihapus: {data}")

# d ubah pisang menjadi nanas
data[4] = "nanas"
print(f"setelah diubah : {data}")

# e tampilkan 3 buah pertama
tiga_buah_pertama = data[:3]
print("Tiga buah pertama:", tiga_buah_pertama)

# no 2
data_angka = [45,12,78,23,56,89,34]

# a urutkan list secara ascendig
data_angka.sort()
print(f"setelah diurut : {data_angka}")

# b urutkan descending
data_angka.reverse()
print(f"setelah dibalik : {data_angka}")
```
