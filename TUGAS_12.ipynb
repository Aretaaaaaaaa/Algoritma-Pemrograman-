```python
#  a luas persegi panjang
def hitung_luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

print("Luas Persegi Panjang")
panjang_persegi = int(input("Masukkan panjang persegi panjang: "))
lebar_persegi = int(input("Masukkan lebar persegi panjang: "))

luas = hitung_luas_persegi_panjang(panjang_persegi, lebar_persegi)
print(f"Luas persegi panjang adalah: {luas}")
```

```python
 # b volume balok
def hitung_volume_balok(panjang, lebar,tinggi):
    return panjang * lebar * tinggi

print("volume balok")
panjang_balok = int(input("Masukkan panjang balok: "))
lebar_balok = int(input("Masukkan lebar balok: "))
tinggi_balok = int(input("Masukkan tinggi balok: "))

volume = hitung_volume_balok(panjang_balok, lebar_balok, tinggi_balok)
print(f"volume balok: {volume}")
```

```python
# C luas lingkaran
def hitung_luas_lingkaran(jari_jari):
    return 3.14 * jari_jari**2

print("luas lingkaran")
jari_jari = int(input("Jari-jari :"))

luas = hitung_luas_lingkaran(jari_jari)
print(f"luas lingkaran {luas}")
```

```python
# D volume balok
def hitung_volume_bola(jari_jari):
    return 75 * 3.14 * jari_jari**3

print("volume bola")
jari_jari = int(input("Jari-jari\t:"))

volume = hitung_volume_bola(jari_jari)
print(f"volume balok {volume}")
```
