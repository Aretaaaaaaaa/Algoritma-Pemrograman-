```python
# a. Buat dictionary keranjang dengan data: {"apel": 5000, "pisang": 3000, "jeruk": 7000,"anggur": 15000}
keranjang = {
    "apel": 5000,
    "pisang": 3000,
    "jeruk": 7000,
    "anggur": 15000
}
# b. Menghitung total harga semua barang menggunakan perulangan
total_harga = 0
for buah, harga in keranjang.items():
    total_harga += harga

# c. Menampilkan setiap barang dan harganya
    print(f"Harga {buah}: Rp {harga}")

# d. Menampilkan harga total belanja
print("\nTotal harga belanja: Rp", total_harga)
```
